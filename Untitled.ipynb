```python
from pathlib import Path
import torch
from PIL import Image
import matplotlib.pyplot as plt

data_dir = Path(
    r"C:\chromosome_ai\chromosome_stage1_data\chromosome_stage1_data\CHROMA\open_data\sta_ab"
)

normal = sorted(data_dir.glob("*/normal/*.png"))
abnormal = sorted(data_dir.glob("*/abnormal/*.png"))

print("정상 사진:", len(normal), "장")
print("이상 사진:", len(abnormal), "장")
print("GPU:", torch.cuda.get_device_name(0))
```

```python
import pandas as pd

records = []

for label, files in [("normal", normal), ("abnormal", abnormal)]:
    for path in files:
        chromosome = (
            path.parent.parent.name
            .removeprefix("sta_dist_")
            .removesuffix("_2")
        )

        records.append({
            "path": str(path),
            "chromosome": chromosome,
            "label": label
        })

df = pd.DataFrame(records)

summary = pd.crosstab(df["chromosome"], df["label"])
summary = summary.reindex(
    [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
)

print("전체 사진:", len(df), "장")
print("\n번호별 사진 수:")
print(summary.to_string())
```

```python
from sklearn.model_selection import train_test_split

df = df.copy()

# 정상=0, 이상=1
df["target"] = df["label"].map({"normal": 0, "abnormal": 1})

# 염색체 번호와 정상/이상 비율을 함께 유지
df["stratum"] = df["chromosome"] + "_" + df["label"]

train_val_df, test_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df["stratum"],
    random_state=42
)

train_df, val_df = train_test_split(
    train_val_df,
    test_size=0.25,
    stratify=train_val_df["stratum"],
    random_state=42
)

# 같은 파일 경로가 여러 용도에 들어가지 않았는지 확인
train_paths = set(train_df["path"])
val_paths = set(val_df["path"])
test_paths = set(test_df["path"])

assert train_paths.isdisjoint(val_paths)
assert train_paths.isdisjoint(test_paths)
assert val_paths.isdisjoint(test_paths)

split_dir = Path(r"C:\chromosome_ai\splits\chroma_demo_v1")
split_dir.mkdir(parents=True, exist_ok=True)

for name, subset in [
    ("train", train_df),
    ("validation", val_df),
    ("test", test_df)
]:
    subset.to_csv(
        split_dir / f"{name}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(
        f"{name}: {len(subset)}장",
        f"| 정상 {(subset['target'] == 0).sum()}장",
        f"| 이상 {(subset['target'] == 1).sum()}장"
    )

print("\n사진 목록 저장 위치:", split_dir)
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image

torch.manual_seed(42)

# 사진 크기·채널·픽셀값을 모델 입력에 맞게 변환
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class ChromosomeDataset(Dataset):
    def __init__(self, table, transform):
        self.table = table.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.table)

    def __getitem__(self, index):
        row = self.table.iloc[index]

        with Image.open(row["path"]) as img:
            image = self.transform(img.convert("RGB"))

        target = int(row["target"])
        return image, target

train_dataset = ChromosomeDataset(train_df, transform)
val_dataset = ChromosomeDataset(val_df, transform)
test_dataset = ChromosomeDataset(test_df, transform)

# 윈도우 주피터에서는 우선 num_workers=0으로 사용
train_loader = DataLoader(
    train_dataset, batch_size=16, shuffle=True, num_workers=0
)
val_loader = DataLoader(
    val_dataset, batch_size=16, shuffle=False, num_workers=0
)
test_loader = DataLoader(
    test_dataset, batch_size=16, shuffle=False, num_workers=0
)

images, labels = next(iter(train_loader))

print("사진 묶음 크기:", images.shape)
print("정답 묶음 크기:", labels.shape)
print("정답 예시:", labels.tolist())
print("정답 의미: 0=정상, 1=이상")
print("학습용 묶음 수:", len(train_loader))
```

```python
import torch
import torch.nn as nn
from torchvision.models import resnet18, ResNet18_Weights

torch.manual_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 일반 이미지로 사전학습된 모델 불러오기
# 처음 실행할 때 모델 파일이 다운로드돼
model = resnet18(weights=ResNet18_Weights.IMAGENET1K_V1)

# 기존 특징 추출 부분 고정
for parameter in model.parameters():
    parameter.requires_grad = False

# 마지막 출력 부분 교체: 0=정상, 1=이상
model.fc = nn.Linear(model.fc.in_features, 2)
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.fc.parameters(), lr=0.001)

# GPU에서 사진 한 묶음을 처리할 수 있는지 확인
model.eval()

with torch.no_grad():
    outputs = model(images.to(device))

print("사용 장치:", device)
print("출력 크기:", outputs.shape)
print(
    "학습할 파라미터 수:",
    sum(p.numel() for p in model.parameters() if p.requires_grad)
)
print("모델 준비 완료 — 아직 학습 전")
```

```python
from tqdm.auto import tqdm

def run_epoch(loader, training=False):
    # 고정된 특징 추출 부분의 설정 유지
    model.eval()
    model.fc.train(training)

    total_loss = 0.0
    correct = 0
    total = 0

    description = "학습" if training else "검증"

    with torch.set_grad_enabled(training):
        for batch_images, batch_labels in tqdm(
            loader, desc=description
        ):
            batch_images = batch_images.to(device)
            batch_labels = batch_labels.to(device)

            if training:
                optimizer.zero_grad(set_to_none=True)

            outputs = model(batch_images)
            loss = criterion(outputs, batch_labels)

            if training:
                loss.backward()
                optimizer.step()

            count = batch_labels.size(0)
            total_loss += loss.item() * count
            correct += (
                outputs.argmax(dim=1) == batch_labels
            ).sum().item()
            total += count

    return total_loss / total, correct / total


# 같은 셀을 다시 실행하면 다음 에포크로 이어서 진행
if "history" not in globals():
    history = []
    best_val_loss = float("inf")

epoch = len(history) + 1

train_loss, train_acc = run_epoch(train_loader, training=True)
val_loss, val_acc = run_epoch(val_loader, training=False)

history.append({
    "epoch": epoch,
    "train_loss": train_loss,
    "train_acc": train_acc,
    "val_loss": val_loss,
    "val_acc": val_acc
})

save_dir = Path(r"C:\chromosome_ai\models\resnet18_head_v1")
save_dir.mkdir(parents=True, exist_ok=True)

improved = val_loss < best_val_loss
if improved:
    best_val_loss = val_loss

checkpoint = {
    "epoch": epoch,
    "model_state_dict": model.state_dict(),
    "optimizer_state_dict": optimizer.state_dict(),
    "history": history,
    "best_val_loss": best_val_loss,
    "class_to_idx": {"normal": 0, "abnormal": 1},
    "model_name": "resnet18",
    "split_directory": str(split_dir)
}

torch.save(checkpoint, save_dir / "last.pth")

if improved:
    torch.save(checkpoint, save_dir / "best.pth")

print(f"\n===== {epoch}에포크 결과 =====")
print(f"학습 Loss: {train_loss:.4f} | 정확도: {train_acc:.2%}")
print(f"검증 Loss: {val_loss:.4f} | 정확도: {val_acc:.2%}")
print("모델 저장 위치:", save_dir)
```

```python
# 현재까지 완료한 다음 에포크부터 총 10에포크까지 진행
for epoch in range(len(history) + 1, 11):

    train_loss, train_acc = run_epoch(
        train_loader, training=True
    )
    val_loss, val_acc = run_epoch(
        val_loader, training=False
    )

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc
    })

    improved = val_loss < best_val_loss

    if improved:
        best_val_loss = val_loss

    checkpoint = {
        "epoch": epoch,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "history": history,
        "best_val_loss": best_val_loss,
        "class_to_idx": {"normal": 0, "abnormal": 1},
        "model_name": "resnet18",
        "split_directory": str(split_dir),
        "preprocessing": {
            "resize": [224, 224],
            "grayscale_channels": 3,
            "mean": [0.485, 0.456, 0.406],
            "std": [0.229, 0.224, 0.225]
        }
    }

    torch.save(checkpoint, save_dir / "last.pth")

    if improved:
        torch.save(checkpoint, save_dir / "best.pth")

    pd.DataFrame(history).to_csv(
        save_dir / "history.csv",
        index=False,
        encoding="utf-8-sig"
    )

    mark = " ★ 최적 모델 저장" if improved else ""

    print(
        f"\nEpoch {epoch}/10"
        f" | 학습 Loss {train_loss:.4f}"
        f" / 정확도 {train_acc:.2%}"
        f" | 검증 Loss {val_loss:.4f}"
        f" / 정확도 {val_acc:.2%}"
        f"{mark}"
    )
```

```python
from sklearn.metrics import confusion_matrix, classification_report

# 검증 Loss가 가장 낮았던 모델 불러오기
best_checkpoint = torch.load(
    save_dir / "best.pth",
    map_location=device,
    weights_only=True
)

model.load_state_dict(best_checkpoint["model_state_dict"])
model.eval()

true_labels = []
pred_labels = []

with torch.no_grad():
    for batch_images, batch_labels in val_loader:
        outputs = model(batch_images.to(device))
        predictions = outputs.argmax(dim=1).cpu()

        true_labels.extend(batch_labels.tolist())
        pred_labels.extend(predictions.tolist())

matrix = confusion_matrix(
    true_labels, pred_labels, labels=[0, 1]
)

print("불러온 모델:", best_checkpoint["epoch"], "에포크")
print("\n행=실제 정답 / 열=모델 예측")

display(pd.DataFrame(
    matrix,
    index=["실제 정상", "실제 이상"],
    columns=["예측 정상", "예측 이상"]
))

print("\n분류 결과:")
print(classification_report(
    true_labels,
    pred_labels,
    labels=[0, 1],
    target_names=["Normal", "Abnormal"],
    digits=3,
    zero_division=0
))
```

```python
# 1. 기존 실험의 최적 모델(5에포크)에서 시작
checkpoint = torch.load(
    r"C:\chromosome_ai\models\resnet18_head_v1\best.pth",
    map_location=device,
    weights_only=True
)
model.load_state_dict(checkpoint["model_state_dict"])

# 2. 전체를 고정한 뒤 layer4와 분류층만 학습 허용
for parameter in model.parameters():
    parameter.requires_grad = False

for parameter in model.layer4.parameters():
    parameter.requires_grad = True

for parameter in model.fc.parameters():
    parameter.requires_grad = True

# 3. 새 optimizer: 기존 특징은 조금씩, 분류층은 더 크게 조정
optimizer = torch.optim.Adam(
    [
        {"params": model.layer4.parameters(), "lr": 1e-5},
        {"params": model.fc.parameters(), "lr": 1e-4},
    ],
    weight_decay=1e-4
)

# 4. 새 실험의 결과를 별도로 저장
save_dir = Path(r"C:\chromosome_ai\models\resnet18_finetune_v1")
save_dir.mkdir(parents=True, exist_ok=True)

history = []
best_val_loss = float("inf")

trainable = sum(
    parameter.numel()
    for parameter in model.parameters()
    if parameter.requires_grad
)

print("시작 모델: 기존 실험의 5에포크 모델")
print(f"학습 가능한 파라미터: {trainable:,}개")
print("저장 폴더:", save_dir)
print("미세조정 준비 완료 — 아직 학습은 시작하지 않았어.")
```

```python
# 출발 모델의 검증 성능 확인
baseline_loss, baseline_acc = run_epoch(val_loader, training=False)
best_val_loss = baseline_loss
history = []

def save_finetune_checkpoint(path, epoch):
    torch.save(
        {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "history": history,
            "best_val_loss": best_val_loss,
            "class_to_idx": {"normal": 0, "abnormal": 1},
            "model_name": "resnet18",
            "trainable_layers": ["layer4", "fc"],
            "split_directory": str(split_dir),
            "preprocessing": {
                "resize": [224, 224],
                "grayscale_channels": 3,
                "mean": [0.485, 0.456, 0.406],
                "std": [0.229, 0.224, 0.225],
            },
        },
        path,
    )

# 미세조정 0에포크 = 기존 모델 그대로
save_finetune_checkpoint(save_dir / "best.pth", epoch=0)

print(
    f"출발 성능 | 검증 Loss {baseline_loss:.4f}"
    f" / 정확도 {baseline_acc:.2%}\n"
)

for epoch in range(1, 11):
    train_loss, train_acc = run_epoch(train_loader, training=True)
    val_loss, val_acc = run_epoch(val_loader, training=False)

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
    })

    improved = val_loss < best_val_loss

    if improved:
        best_val_loss = val_loss
        save_finetune_checkpoint(save_dir / "best.pth", epoch)

    save_finetune_checkpoint(save_dir / "last.pth", epoch)
    pd.DataFrame(history).to_csv(
        save_dir / "history.csv",
        index=False,
        encoding="utf-8-sig",
    )

    mark = " ★ 최적 모델 저장" if improved else ""
    print(
        f"\n미세조정 Epoch {epoch}/10 | "
        f"학습 Loss {train_loss:.4f} / 정확도 {train_acc:.2%} | "
        f"검증 Loss {val_loss:.4f} / 정확도 {val_acc:.2%}"
        f"{mark}"
    )

print("\n학습 완료")
print("저장 위치:", save_dir)
```

```python
import numpy as np

def crop_chromosome_preview(image):
    gray = image.convert("L")
    array = np.asarray(gray)

    # 어두운 픽셀로 대략적인 범위만 찾음
    # 실제 염색체 분할이나 이상 위치 검출은 아님
    foreground = array < 240
    ys, xs = np.where(foreground)

    if len(xs) == 0:
        return ImageOps.pad(gray, (224, 224), color=255)

    left, right = int(xs.min()), int(xs.max()) + 1
    top, bottom = int(ys.min()), int(ys.max()) + 1

    # 가장자리가 잘리지 않도록 여유 공간 추가
    margin = max(4, round(max(right - left, bottom - top) * 0.15))

    box = (
        max(0, left - margin),
        max(0, top - margin),
        min(gray.width, right + margin),
        min(gray.height, bottom + margin),
    )

    cropped = gray.crop(box)

    return ImageOps.pad(
        cropped,
        (224, 224),
        method=Image.Resampling.BILINEAR,
        color=255,
    )

# 학습 데이터에서 정상 4장, 이상 4장 선택
preview_samples = pd.concat([
    train_df[train_df["target"] == 0].sample(4, random_state=42),
    train_df[train_df["target"] == 1].sample(4, random_state=42),
])

fig, axes = plt.subplots(4, 4, figsize=(12, 12))

for index, (_, item) in enumerate(preview_samples.iterrows()):
    row = index // 2
    column = (index % 2) * 2

    with Image.open(item["path"]) as source:
        original = source.convert("L")

    cropped = crop_chromosome_preview(original)
    label_name = "Normal" if item["target"] == 0 else "Abnormal"

    for offset, (picture, name) in enumerate([
        (original, "Original"),
        (cropped, "Crop + padding"),
    ]):
        ax = axes[row, column + offset]
        ax.imshow(picture, cmap="gray", vmin=0, vmax=255)
        ax.set_title(
            f"{item['chromosome']} | {label_name}\n{name}",
            fontsize=9
        )
        ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
# 1. 여백 제거 전처리
crop_transform = transforms.Compose([
    transforms.Lambda(crop_chromosome_preview),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225],
    ),
])

# 2. 기존 분할을 그대로 사용
crop_train_dataset = ChromosomeDataset(train_df, crop_transform)
crop_val_dataset = ChromosomeDataset(val_df, crop_transform)

crop_train_loader = DataLoader(
    crop_train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=0,
    generator=torch.Generator().manual_seed(42),
)

crop_val_loader = DataLoader(
    crop_val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0,
)

# 3. ImageNet 사전학습 모델에서 새로 시작
# 기존 실험과 마찬가지로 마지막 분류층만 학습
torch.manual_seed(42)

crop_model = resnet18(
    weights=ResNet18_Weights.IMAGENET1K_V1
)

for parameter in crop_model.parameters():
    parameter.requires_grad = False

crop_model.fc = nn.Linear(crop_model.fc.in_features, 2)
crop_model = crop_model.to(device)

crop_optimizer = torch.optim.Adam(
    crop_model.fc.parameters(),
    lr=0.001,
)

# 4. 별도 저장 폴더
crop_save_dir = Path(
    r"C:\chromosome_ai\models\resnet18_crop_head_v1"
)
crop_save_dir.mkdir(parents=True, exist_ok=True)

sample_image, sample_label = crop_train_dataset[0]

print("학습 사진:", len(crop_train_dataset), "장")
print("검증 사진:", len(crop_val_dataset), "장")
print("입력 크기:", sample_image.shape)
print(
    "학습 가능한 파라미터:",
    sum(p.numel() for p in crop_model.parameters() if p.requires_grad),
)
print("저장 폴더:", crop_save_dir)
print("여백 제거 실험 준비 완료 — 아직 학습 전")
```

```python
from tqdm.auto import tqdm

crop_criterion = nn.CrossEntropyLoss()
crop_history = []
crop_best_loss = float("inf")

def run_crop_epoch(loader, training=False):
    # 고정된 특징 추출부와 BatchNorm은 평가 모드 유지
    crop_model.eval()
    crop_model.fc.train(training)

    total_loss = 0.0
    correct = 0
    total = 0
    abnormal_correct = 0
    abnormal_total = 0

    with torch.set_grad_enabled(training):
        for batch_images, batch_labels in tqdm(
            loader,
            desc="학습" if training else "검증"
        ):
            batch_images = batch_images.to(device)
            batch_labels = batch_labels.to(device)

            if training:
                crop_optimizer.zero_grad(set_to_none=True)

            outputs = crop_model(batch_images)
            loss = crop_criterion(outputs, batch_labels)

            if training:
                loss.backward()
                crop_optimizer.step()

            predictions = outputs.argmax(dim=1)
            abnormal_mask = batch_labels == 1

            count = batch_labels.size(0)
            total_loss += loss.item() * count
            correct += (predictions == batch_labels).sum().item()
            total += count

            abnormal_correct += (
                (predictions == 1) & abnormal_mask
            ).sum().item()
            abnormal_total += abnormal_mask.sum().item()

    return (
        total_loss / total,
        correct / total,
        abnormal_correct / max(abnormal_total, 1),
    )


for epoch in range(1, 11):
    train_loss, train_acc, _ = run_crop_epoch(
        crop_train_loader, training=True
    )
    val_loss, val_acc, val_recall = run_crop_epoch(
        crop_val_loader, training=False
    )

    crop_history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "val_abnormal_recall": val_recall,
    })

    improved = val_loss < crop_best_loss
    if improved:
        crop_best_loss = val_loss

    checkpoint = {
        "epoch": epoch,
        "model_state_dict": crop_model.state_dict(),
        "optimizer_state_dict": crop_optimizer.state_dict(),
        "history": crop_history,
        "best_val_loss": crop_best_loss,
        "class_to_idx": {"normal": 0, "abnormal": 1},
        "model_name": "resnet18",
        "trainable_layers": ["fc"],
        "split_directory": str(split_dir),
        "preprocessing": {
            "crop_function": "crop_chromosome_preview",
            "foreground_threshold": 240,
            "margin_fraction": 0.15,
            "minimum_margin_pixels": 4,
            "output_size": [224, 224],
            "keep_aspect_ratio": True,
            "padding_color": 255,
            "resize_interpolation": "bilinear",
            "grayscale_channels": 3,
            "mean": [0.485, 0.456, 0.406],
            "std": [0.229, 0.224, 0.225],
        },
    }

    if improved:
        torch.save(checkpoint, crop_save_dir / "best.pth")

    torch.save(checkpoint, crop_save_dir / "last.pth")

    pd.DataFrame(crop_history).to_csv(
        crop_save_dir / "history.csv",
        index=False,
        encoding="utf-8-sig",
    )

    mark = " ★ 최적 모델 저장" if improved else ""
    print(
        f"\nEpoch {epoch}/10 | "
        f"학습 Loss {train_loss:.4f} / 정확도 {train_acc:.2%} | "
        f"검증 Loss {val_loss:.4f} / 정확도 {val_acc:.2%} | "
        f"이상 재현율 {val_recall:.2%}{mark}"
    )

print("\n학습 완료")
print("저장 위치:", crop_save_dir)
```

```python
from sklearn.metrics import (
    roc_auc_score,
    accuracy_score,
    recall_score,
    confusion_matrix,
)

def evaluate_validation(model_to_use, loader, checkpoint_path):
    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=True,
    )
    model_to_use.load_state_dict(checkpoint["model_state_dict"])
    model_to_use.eval()

    targets = []
    scores = []

    with torch.no_grad():
        for batch_images, batch_labels in loader:
            outputs = model_to_use(batch_images.to(device))
            probabilities = torch.softmax(outputs, dim=1)[:, 1]

            targets.extend(batch_labels.tolist())
            scores.extend(probabilities.cpu().tolist())

    predictions = (np.array(scores) >= 0.5).astype(int)
    tn, fp, fn, tp = confusion_matrix(
        targets, predictions, labels=[0, 1]
    ).ravel()

    return {
        "ROC-AUC": roc_auc_score(targets, scores),
        "정확도": accuracy_score(targets, predictions),
        "이상 재현율": recall_score(targets, predictions),
        "정상 재현율": tn / (tn + fp),
        "정상→이상": int(fp),
        "이상→정상": int(fn),
    }

comparison = pd.DataFrame({
    "기존 모델": evaluate_validation(
        model,
        val_loader,
        Path(r"C:\chromosome_ai\models\resnet18_head_v1\best.pth"),
    ),
    "여백 제거 모델": evaluate_validation(
        crop_model,
        crop_val_loader,
        crop_save_dir / "best.pth",
    ),
}).T

display(comparison.round(4))
```

```python
from sklearn.metrics import roc_auc_score
from tqdm.auto import tqdm

aug_criterion = nn.CrossEntropyLoss()

def run_aug_epoch(loader, training=False):
    # 작은 데이터셋이므로 BatchNorm 통계는 고정
    aug_model.eval()
    aug_model.fc.train(training)

    total_loss = 0.0
    targets, scores = [], []

    with torch.set_grad_enabled(training):
        for batch_images, batch_labels in tqdm(
            loader, desc="학습" if training else "검증"
        ):
            batch_images = batch_images.to(device)
            batch_labels = batch_labels.to(device)

            if training:
                aug_optimizer.zero_grad(set_to_none=True)

            outputs = aug_model(batch_images)
            loss = aug_criterion(outputs, batch_labels)

            if training:
                loss.backward()
                aug_optimizer.step()

            total_loss += loss.item() * batch_labels.size(0)
            targets.extend(batch_labels.cpu().tolist())
            scores.extend(
                torch.softmax(outputs.detach(), dim=1)[:, 1]
                .cpu().tolist()
            )

    targets = np.array(targets)
    scores = np.array(scores)
    predictions = (scores >= 0.5).astype(int)

    return {
        "loss": total_loss / len(targets),
        "acc": float((predictions == targets).mean()),
        "recall": float((predictions[targets == 1] == 1).mean()),
        "auc": float(roc_auc_score(targets, scores)),
    }


aug_history = []
aug_best_loss = float("inf")
aug_no_improvement = 0

def save_aug_checkpoint(path, epoch):
    torch.save({
        "epoch": epoch,
        "model_state_dict": aug_model.state_dict(),
        "optimizer_state_dict": aug_optimizer.state_dict(),
        "history": aug_history,
        "best_val_loss": aug_best_loss,
        "model_name": "resnet18",
        "trainable_layers": ["layer4", "fc"],
        "class_to_idx": {"normal": 0, "abnormal": 1},
        "split_directory": str(split_dir),
        "preprocessing": {
            "resize": [224, 224],
            "grayscale_channels": 3,
            "mean": [0.485, 0.456, 0.406],
            "std": [0.229, 0.224, 0.225],
        },
        "training_augmentation": {
            "rotation_degrees": 10,
            "translate_fraction": [0.03, 0.03],
            "interpolation": "bilinear",
            "fill": 255,
        },
        "threshold": 0.5,
    }, path)


# 0에포크: 출발 모델 보존
baseline = run_aug_epoch(aug_val_loader)
aug_best_loss = baseline["loss"]
save_aug_checkpoint(aug_save_dir / "best.pth", epoch=0)
save_aug_checkpoint(aug_save_dir / "last.pth", epoch=0)

print(
    f"\n출발 모델 | Loss {baseline['loss']:.4f} | "
    f"정확도 {baseline['acc']:.2%} | "
    f"이상 재현율 {baseline['recall']:.2%} | "
    f"ROC-AUC {baseline['auc']:.4f}"
)

for epoch in range(1, 11):
    train_metrics = run_aug_epoch(aug_train_loader, training=True)
    val_metrics = run_aug_epoch(aug_val_loader)

    aug_history.append({
        "epoch": epoch,
        "train_loss": train_metrics["loss"],
        "train_acc": train_metrics["acc"],
        **{f"val_{key}": value for key, value in val_metrics.items()},
    })

    improved = val_metrics["loss"] < aug_best_loss

    if improved:
        aug_best_loss = val_metrics["loss"]
        aug_no_improvement = 0
        save_aug_checkpoint(aug_save_dir / "best.pth", epoch)
    else:
        aug_no_improvement += 1

    save_aug_checkpoint(aug_save_dir / "last.pth", epoch)
    pd.DataFrame(aug_history).to_csv(
        aug_save_dir / "history.csv",
        index=False,
        encoding="utf-8-sig",
    )

    mark = " ★ 최적 모델 저장" if improved else ""
    print(
        f"\nEpoch {epoch}/10 | "
        f"학습 Loss {train_metrics['loss']:.4f} / "
        f"정확도 {train_metrics['acc']:.2%} | "
        f"검증 Loss {val_metrics['loss']:.4f} / "
        f"정확도 {val_metrics['acc']:.2%} | "
        f"이상 재현율 {val_metrics['recall']:.2%} | "
        f"ROC-AUC {val_metrics['auc']:.4f}{mark}"
    )

    if aug_no_improvement >= 3:
        print("\n검증 Loss가 3에포크 연속 개선되지 않아 중단했어.")
        break

print("\n완료. 저장 위치:", aug_save_dir)
```

```python
import timm
import torch

# 1. 분류층이 없는 ViT-Large 인코더 생성
chroma_encoder = timm.create_model(
    "vit_large_patch16_224",
    pretrained=False,
    num_classes=0,
    global_pool="token",
)

# 2. 복원 학습에만 필요한 가중치 제외
encoder_state = {
    name: value
    for name, value in chroma_state.items()
    if not name.startswith("decoder_")
    and name != "mask_token"
}

# 모든 인코더 가중치가 정확히 맞는지 검사하며 로딩
load_result = chroma_encoder.load_state_dict(
    encoder_state,
    strict=True,
)
print("가중치 로딩:", load_result)

# 3. 특징 추출부 고정
for parameter in chroma_encoder.parameters():
    parameter.requires_grad = False

chroma_encoder = chroma_encoder.to(device)
chroma_encoder.eval()

# 4. 기존 전처리로 학습 사진 한 장만 시험
sample_image, sample_target = train_dataset[0]
sample_batch = sample_image.unsqueeze(0).to(device)

with torch.inference_mode():
    sample_features = chroma_encoder(sample_batch)

print("입력 크기:", tuple(sample_batch.shape))
print("특징 벡터 크기:", tuple(sample_features.shape))
print(
    "특징값 모두 정상:",
    torch.isfinite(sample_features).all().item(),
)
print("실행 장치:", sample_features.device)

if device.type == "cuda":
    print(
        "현재 PyTorch GPU 할당 메모리:",
        f"{torch.cuda.memory_allocated() / 1024**3:.2f} GiB",
    )

print("특징 추출 시험 완료 — 아직 분류 학습 전")
```

```python
from torch.utils.data import DataLoader
from pathlib import Path
from tqdm.auto import tqdm

feature_dir = Path(
    r"C:\chromosome_ai\features\chroma_pretrain_token_v1"
)
feature_dir.mkdir(parents=True, exist_ok=True)

feature_metadata = {
    "model": "vit_large_patch16_224",
    "repository": "Boyoungc/CHROMA",
    "revision": "887f65b56fc93cec08b9130f19bfbd45f1fa7ac8",
    "checkpoint": str(chroma_weight_path),
    "pooling": "token",
    "feature_dimension": 1024,
    "preprocessing": str(train_dataset.transform),
    "class_to_idx": {"normal": 0, "abnormal": 1},
    "split_directory": str(split_dir),
}

def extract_and_save_features(dataset, split_name):
    loader = DataLoader(
        dataset,
        batch_size=4,
        shuffle=False,
        num_workers=0,
    )

    feature_batches = []
    label_batches = []

    chroma_encoder.eval()

    with torch.inference_mode():
        for batch_images, batch_labels in tqdm(
            loader, desc=f"{split_name} 특징 추출"
        ):
            features = chroma_encoder(batch_images.to(device))

            feature_batches.append(features.cpu())
            label_batches.append(batch_labels.cpu())

    all_features = torch.cat(feature_batches, dim=0)
    all_labels = torch.cat(label_batches, dim=0)

    assert all_features.shape == (len(dataset), 1024)
    assert torch.isfinite(all_features).all().item()

    payload = {
        "features": all_features,
        "labels": all_labels,
        "paths": dataset.table["path"].astype(str).tolist(),
        "chromosomes": dataset.table["chromosome"].astype(str).tolist(),
        "metadata": feature_metadata,
    }

    output_path = feature_dir / f"{split_name}_features.pt"
    torch.save(payload, output_path)

    print(f"\n{split_name}: {tuple(all_features.shape)}")
    print("저장 위치:", output_path)

    return payload


chroma_train_features = extract_and_save_features(
    train_dataset, "train"
)

chroma_val_features = extract_and_save_features(
    val_dataset, "validation"
)

print("\n학습·검증 특징 저장 완료 — 아직 분류층 학습 전")
```

```python
import joblib
import numpy as np
import pandas as pd

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    recall_score,
    roc_auc_score,
    log_loss,
    confusion_matrix,
)

# 1. 저장된 특징과 정답 준비
X_train = chroma_train_features["features"].numpy()
y_train = chroma_train_features["labels"].numpy()

X_val = chroma_val_features["features"].numpy()
y_val = chroma_val_features["labels"].numpy()

# 2. 특징 표준화 + 정상/이상 선형 분류
chroma_classifier = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        C=1.0,
        max_iter=3000,
        solver="lbfgs",
        random_state=42,
    ),
)

chroma_classifier.fit(X_train, y_train)

# 3. 학습·검증 성능 확인
def evaluate_chroma_classifier(features, targets):
    abnormal_column = list(chroma_classifier.classes_).index(1)
    scores = chroma_classifier.predict_proba(features)[:, abnormal_column]
    predictions = (scores >= 0.5).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        targets, predictions, labels=[0, 1]
    ).ravel()

    metrics = {
        "Loss": log_loss(targets, scores, labels=[0, 1]),
        "정확도": accuracy_score(targets, predictions),
        "ROC-AUC": roc_auc_score(targets, scores),
        "이상 재현율": recall_score(targets, predictions),
        "정상 재현율": tn / (tn + fp),
        "정상 오탐": int(fp),
        "이상 놓침": int(fn),
    }
    return metrics, scores, predictions

train_metrics, _, _ = evaluate_chroma_classifier(X_train, y_train)
val_metrics, chroma_val_scores, chroma_val_predictions = (
    evaluate_chroma_classifier(X_val, y_val)
)

display(pd.DataFrame(
    [train_metrics, val_metrics],
    index=["학습", "검증"],
).round(4))

print("\n검증 혼동행렬:")
display(pd.DataFrame(
    confusion_matrix(y_val, chroma_val_predictions, labels=[0, 1]),
    index=["실제 정상", "실제 이상"],
    columns=["예측 정상", "예측 이상"],
))

# 4. 분류기와 실험 정보 저장
chroma_classifier_dir = Path(
    r"C:\chromosome_ai\models\chroma_linear_v1"
)
chroma_classifier_dir.mkdir(parents=True, exist_ok=True)

joblib.dump(
    {
        "classifier": chroma_classifier,
        "feature_metadata": feature_metadata,
        "threshold": 0.5,
        "train_metrics": train_metrics,
        "validation_metrics": val_metrics,
    },
    chroma_classifier_dir / "classifier.joblib",
)

print("\n분류기 학습·저장 완료")
print("저장 위치:", chroma_classifier_dir)
```

```python
# 기존 테스트 사진의 CHROMA 특징 추출
# 이 결과는 비교 참고용이며, 새 외부 데이터 평가는 아님
chroma_test_reference = extract_and_save_features(
    test_dataset,
    "test_reference",
)

X_reference = chroma_test_reference["features"].numpy()
y_reference = chroma_test_reference["labels"].numpy()

reference_metrics, reference_scores, reference_predictions = (
    evaluate_chroma_classifier(X_reference, y_reference)
)

print("기존 테스트 96장에 대한 참고 비교:")
display(pd.DataFrame(
    [reference_metrics],
    index=["CHROMA 참고 평가"],
).round(4))

display(pd.DataFrame(
    confusion_matrix(
        y_reference, reference_predictions, labels=[0, 1]
    ),
    index=["실제 정상", "실제 이상"],
    columns=["예측 정상", "예측 이상"],
))

# 사진별 결과 저장
reference_results = test_dataset.table.copy()
reference_results["abnormal_score"] = reference_scores
reference_results["prediction"] = reference_predictions

reference_results.to_csv(
    chroma_classifier_dir / "test_reference_predictions.csv",
    index=False,
    encoding="utf-8-sig",
)

pd.DataFrame([reference_metrics]).to_csv(
    chroma_classifier_dir / "test_reference_metrics.csv",
    index=False,
    encoding="utf-8-sig",
)

print("참고 평가 결과 저장 완료")
```

```python
from PIL import Image
from pathlib import Path

# 기존 학습·검증에 사용한 전처리 유지
inference_transform = train_dataset.transform

def predict_chromosome(image_path):
    image_path = Path(image_path)

    if not image_path.is_file():
        raise FileNotFoundError(f"사진을 찾을 수 없어: {image_path}")

    with Image.open(image_path) as source:
        image = source.convert("RGB")

    inputs = inference_transform(image).unsqueeze(0).to(device)

    chroma_encoder.eval()
    chroma_number_encoder.eval()

    # 두 분류기는 서로 다른 인코더의 특징을 사용
    with torch.inference_mode():
        status_features = chroma_encoder(inputs).cpu().numpy()
        number_features = chroma_number_encoder(inputs).cpu().numpy()

    # 정상·이상 결과
    status_probabilities = chroma_classifier.predict_proba(
        status_features
    )[0]
    abnormal_column = list(chroma_classifier.classes_).index(1)
    abnormal_score = float(status_probabilities[abnormal_column])

    # 번호 후보 상위 3개
    number_scores = number_task_classifier.predict_proba(
        number_features
    )[0]
    top_indices = np.argsort(number_scores)[::-1][:3]

    candidates = []
    for index in top_indices:
        class_id = int(number_task_classifier.classes_[index])
        candidates.append({
            "번호 후보": chromosome_names[class_id],
            "모델 점수(%)": round(float(number_scores[index]) * 100, 2),
        })

    return {
        "image_path": str(image_path),
        "number_candidates": candidates,
        "status": "이상 의심" if abnormal_score >= 0.5 else "정상 예측",
        "abnormal_score": abnormal_score,
    }


# 검증 데이터의 이상 사진 한 장으로 화면 시연
# 새로운 데이터에 대한 성능 평가가 아닌 동작 확인용
demo_row = val_dataset.table[
    val_dataset.table["target"] == 1
].iloc[0]

demo_result = predict_chromosome(demo_row["path"])

with Image.open(demo_row["path"]) as source:
    plt.imshow(source.convert("L"), cmap="gray")
    plt.axis("off")
    plt.show()

print("번호 후보:")
display(pd.DataFrame(demo_result["number_candidates"]))

print("상태 예측:", demo_result["status"])
print(f"이상 모델 점수: {demo_result['abnormal_score']:.3f}")

print("\n비교용 데이터 라벨:")
print("실제 번호:", demo_row["chromosome"])
print("실제 상태:", demo_row["label"])
```

```python
import argparse
import joblib
import timm
import torch
import numpy as np
import pandas as pd
from torchvision import transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1. 저장된 분류기 복원
status_bundle = joblib.load(
    project_dir / "models/chroma_linear_v1/classifier.joblib"
)
number_bundle = joblib.load(
    project_dir / "models/chroma_number_task_linear_v1/classifier.joblib"
)

chroma_classifier = status_bundle["classifier"]
number_task_classifier = number_bundle["classifier"]
chromosome_names = number_bundle["chromosome_names"]

print("정상·이상 분류기 / 번호 분류기 복원 완료")


# 2. 인코더 복원 함수
def restore_encoder(weight_path, pooling):
    with torch.serialization.safe_globals([argparse.Namespace]):
        checkpoint = torch.load(
            weight_path,
            map_location="cpu",
            weights_only=True,
            mmap=True,
        )

    state = {
        name: value
        for name, value in checkpoint["model"].items()
        if not name.startswith(("decoder_", "head."))
        and name != "mask_token"
    }

    encoder = timm.create_model(
        "vit_large_patch16_224",
        pretrained=False,
        num_classes=0,
        global_pool=pooling,
    )

    encoder.load_state_dict(state, strict=True)
    encoder.requires_grad_(False)
    encoder = encoder.to(device)
    encoder.eval()

    return encoder


# 정상·이상용: token 풀링
chroma_encoder = restore_encoder(
    project_dir / "pretrained/CHROMA/CHROMA_pretrain.pth",
    pooling="token",
)
print("정상·이상 인코더 복원 완료")

# 번호용: 평균 풀링
chroma_number_encoder = restore_encoder(
    project_dir / "pretrained/CHROMA/CHROMA-num_ab.pth",
    pooling="avg",
)
print("번호 인코더 복원 완료")


# 3. 학습 당시와 같은 입력 전처리
inference_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225],
    ),
])

# 4. 시연용 검증 사진 목록 복원
val_df = pd.read_csv(
    project_dir / "splits/chroma_demo_v1/validation.csv"
)

print("\n검증 사진 목록:", len(val_df), "장")
print("실행 장치:", device)
print("모델 복원 완료 — 재학습하지 않았어.")
```

```python
from PIL import Image
import matplotlib.pyplot as plt

def predict_chromosome(image_path):
    image_path = Path(image_path)

    with Image.open(image_path) as source:
        image = source.convert("RGB")

    inputs = inference_transform(image).unsqueeze(0).to(device)

    with torch.inference_mode():
        status_features = chroma_encoder(inputs).cpu().numpy()
        number_features = chroma_number_encoder(inputs).cpu().numpy()

    # 정상·이상 분류
    status_scores = chroma_classifier.predict_proba(status_features)[0]
    abnormal_index = list(chroma_classifier.classes_).index(1)
    abnormal_score = float(status_scores[abnormal_index])
    threshold = status_bundle.get("threshold", 0.5)

    # 번호 후보
    number_scores = number_task_classifier.predict_proba(
        number_features
    )[0]
    top_indices = np.argsort(number_scores)[::-1][:3]

    candidates = []
    for index in top_indices:
        class_id = int(number_task_classifier.classes_[index])
        candidates.append({
            "번호 후보": chromosome_names[class_id],
            "모델 점수(%)": round(float(number_scores[index]) * 100, 2),
        })

    return {
        "image_path": str(image_path),
        "number_candidates": candidates,
        "status": (
            "이상 의심" if abnormal_score >= threshold else "정상 예측"
        ),
        "abnormal_score": abnormal_score,
    }


# 이전과 같이 검증 목록의 첫 번째 이상 사진으로 시연
demo_row = val_df[val_df["target"] == 1].iloc[0]
demo_result = predict_chromosome(demo_row["path"])

with Image.open(demo_result["image_path"]) as source:
    plt.figure(figsize=(3, 3))
    plt.imshow(source.convert("L"), cmap="gray")
    plt.axis("off")
    plt.show()

display(pd.DataFrame(demo_result["number_candidates"]))
print("분석 상태:", demo_result["status"])
print(f"이상 모델 점수: {demo_result['abnormal_score']:.3f}")
print("비교용 실제 라벨:", demo_row["chromosome"], demo_row["label"])
```
