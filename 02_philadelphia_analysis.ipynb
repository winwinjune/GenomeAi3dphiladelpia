```python
분할폴더 = TVG루트 / "splits"

조각목록경로 = {
    "train": 분할폴더 / "tvg_number_652_v1/train_views.csv",
    "validation": 분할폴더 / "number_local_view_v1/validation_views.csv",
    "test": 분할폴더 / "number_local_view_v1/test_final_v1_views.csv",
}

정상조각목록 = {}
사용한원본 = {}

for 분할, 조각경로 in 조각목록경로.items():
    사례 = pd.read_csv(
        분할폴더 / f"tvg_number_652_v1/{분할}_cases.csv"
    )

    정상사례 = 사례.loc[
        사례["source_category"].eq("normal"), ["filename"]
    ].copy()

    assert 정상사례["filename"].is_unique, "사례 목록에 중복 파일명 있음"

    조각 = pd.read_csv(조각경로)

    # 정상 사례의 조각만 연결
    정상조각 = 조각.merge(
        정상사례,
        left_on="source_image",
        right_on="filename",
        how="inner",
        validate="many_to_one",
    )

    # 정상 사례가 조각 목록에서 누락되지 않았는지 검사
    누락원본 = set(정상사례["filename"]) - set(정상조각["source_image"])
    assert not 누락원본, f"{분할}: 조각 목록에 없는 정상 원본 {누락원본}"

    선택 = 정상조각[
        정상조각["chromosome"].isin(["chr9", "chr22"])
    ].copy()

    assert not 선택.empty, f"{분할}: 9번·22번 라벨 확인 필요"
    assert not 선택.duplicated(
        ["source_image", "object_index"]
    ).any(), f"{분할}: 중복 조각 확인 필요"

    정상조각목록[분할] = 선택
    사용한원본[분할] = set(정상사례["filename"])

    print(f"\n=== {분할}: 정상 원본 {len(정상사례)}장 ===")
    print(선택["chromosome"].value_counts())

    for 열 in ["global_crop_path", "local_crop_path"]:
        assert 열 in 선택.columns, f"{분할}: {열} 열 없음"

        def 파일존재(값):
            if pd.isna(값):
                return False
            경로 = Path(str(값))
            if not 경로.is_absolute():
                경로 = TVG루트 / 경로
            return 경로.is_file()

        없는개수 = int((~선택[열].map(파일존재)).sum())
        print(f"{열}: 없는 파일 {없는개수}개")

# 원본 파일명 기준으로 분할 간 중복 검사
for 앞, 뒤 in [
    ("train", "validation"),
    ("train", "test"),
    ("validation", "test"),
]:
    assert not (사용한원본[앞] & 사용한원본[뒤]), \
        f"{앞}/{뒤}에 같은 정상 원본이 있음"

print("\n정상 9번·22번 목록 구성 완료 — 아직 학습하지 않았습니다.")
```

```python
import timm
import pandas as pd
import torch

from torch import nn
from torchvision import transforms
from PIL import Image
from pathlib import Path

전좌용_장치 = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# 기존 모델과 동일한 구조
class 전좌용번호모델(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = timm.create_model(
            "vit_large_patch16_224",
            pretrained=False,
            num_classes=0,
            global_pool="avg",
        )
        self.head = nn.Linear(2048, 24)

    def extract_features(self, 전체, 확대):
        배치크기 = len(전체)
        특징 = self.encoder(torch.cat([전체, 확대], dim=0))
        return torch.cat(
            [특징[:배치크기], 특징[배치크기:]], dim=1
        )

    def forward(self, 전체, 확대):
        return self.head(self.extract_features(전체, 확대))


전좌용_번호모델 = 전좌용번호모델()

# 키와 크기가 모두 일치해야 복원 성공
전좌용_번호모델.load_state_dict(
    전좌용_체크포인트["model"],
    strict=True,
)
전좌용_번호모델.requires_grad_(False)
전좌용_번호모델.eval()
전좌용_번호모델.to(전좌용_장치)

# 저장된 설정과 동일한 전처리
전좌용_전처리 = transforms.Compose([
    transforms.Resize(
        (224, 224),
        interpolation=transforms.InterpolationMode.BILINEAR,
        antialias=True,
    ),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225],
    ),
])

def 전좌용_이미지읽기(값):
    경로 = Path(str(값))
    if not 경로.is_absolute():
        경로 = Path(r"C:\chromosome_ai\tvg_stage2") / 경로
    with Image.open(경로) as 사진:
        return 전좌용_전처리(사진.convert("RGB"))


# 학습 목록에서 번호별 한 개씩만 시험
학습목록 = pd.read_csv(
    Path(r"C:\chromosome_ai\philadelphia_data\manifests")
    / "normal_chr9_chr22_train_v1.csv"
)

시험목록 = pd.concat([
    학습목록[학습목록["chromosome"].eq("chr9")].head(1),
    학습목록[학습목록["chromosome"].eq("chr22")].head(1),
], ignore_index=True)

assert len(시험목록) == 2

전체입력 = torch.stack([
    전좌용_이미지읽기(값)
    for 값 in 시험목록["global_crop_path"]
]).to(전좌용_장치)

확대입력 = torch.stack([
    전좌용_이미지읽기(값)
    for 값 in 시험목록["local_crop_path"]
]).to(전좌용_장치)

with torch.inference_mode():
    시험특징 = 전좌용_번호모델.extract_features(전체입력, 확대입력)
    시험출력 = 전좌용_번호모델.head(시험특징)

assert 시험특징.shape == (2, 2048)
assert torch.isfinite(시험특징).all().item()

번호목록 = 전좌용_체크포인트["chromosome_names"]
시험결과 = 시험목록[
    ["source_image", "object_index", "chromosome"]
].copy()

시험결과["번호예측"] = [
    번호목록[i] for i in 시험출력.argmax(1).cpu().tolist()
]

print("모델 복원 완료")
print("사용 장치:", 전좌용_장치)
print("시험 특징 크기:", tuple(시험특징.shape))
display(시험결과)
print("특징 추출 시험 완료 — 학습·파일 변경 없음")
```

```python
import hashlib
import pandas as pd
import torch
from pathlib import Path
from tqdm.auto import tqdm

프로젝트 = Path(r"C:\chromosome_ai")
목록폴더 = 프로젝트 / "philadelphia_data" / "manifests"
특징폴더 = 프로젝트 / "philadelphia_data" / "features" / "finetuned652_v1"
특징폴더.mkdir(parents=True, exist_ok=True)

def 파일해시(경로):
    해시 = hashlib.sha256()
    with Path(경로).open("rb") as 파일:
        for 조각 in iter(lambda: 파일.read(1024 * 1024), b""):
            해시.update(조각)
    return 해시.hexdigest()

print("모델 파일 식별값 확인 중...")
모델해시 = 파일해시(모델경로)

전좌용_번호모델.eval()
정상특징 = {}

for 분할 in ["train", "validation"]:
    목록경로 = 목록폴더 / f"normal_chr9_chr22_{분할}_v1.csv"
    표 = pd.read_csv(목록경로).reset_index(drop=True)
    저장경로 = 특징폴더 / f"{분할}_features.pt"

    설정 = {
        "checkpoint": str(모델경로),
        "checkpoint_sha256": 모델해시,
        "manifest_sha256": 파일해시(목록경로),
        "preprocessing": str(전좌용_전처리),
        "feature_order": ["global_1024", "local_1024"],
        "feature_dimension": 2048,
        "augmentation": "none",
    }

    # 완료된 결과는 설정과 순서가 일치할 때만 재사용
    if 저장경로.is_file():
        결과 = torch.load(
            저장경로, map_location="cpu", weights_only=True
        )
        assert 결과["metadata"] == 설정, \
            f"{분할}: 기존 특징의 설정이 다릅니다. 덮어쓰지 않습니다."
        assert 결과["source_images"] == 표["source_image"].tolist()
        assert 결과["object_indices"] == 표["object_index"].tolist()
        assert 결과["chromosomes"] == 표["chromosome"].tolist()
        assert 결과["features"].shape == (len(표), 2048)
        assert torch.isfinite(결과["features"]).all().item()

        정상특징[분할] = 결과
        print(f"{분할}: 저장된 특징 재사용 {tuple(결과['features'].shape)}")
        continue

    특징묶음 = []

    with torch.inference_mode():
        for 시작 in tqdm(
            range(0, len(표), 2),
            desc=f"{분할} 특징 추출",
        ):
            배치 = 표.iloc[시작:시작 + 2]

            전체 = torch.stack([
                전좌용_이미지읽기(값)
                for 값 in 배치["global_crop_path"]
            ]).to(전좌용_장치)

            확대 = torch.stack([
                전좌용_이미지읽기(값)
                for 값 in 배치["local_crop_path"]
            ]).to(전좌용_장치)

            특징 = 전좌용_번호모델.extract_features(전체, 확대)
            assert torch.isfinite(특징).all().item()
            특징묶음.append(특징.cpu())

    특징행렬 = torch.cat(특징묶음, dim=0)
    assert 특징행렬.shape == (len(표), 2048)

    결과 = {
        "features": 특징행렬,
        "source_images": 표["source_image"].tolist(),
        "object_indices": 표["object_index"].tolist(),
        "chromosomes": 표["chromosome"].tolist(),
        "global_crop_paths": 표["global_crop_path"].tolist(),
        "local_crop_paths": 표["local_crop_path"].tolist(),
        "metadata": 설정,
    }

    임시경로 = 저장경로.with_suffix(".pt.part")
    torch.save(결과, 임시경로)
    임시경로.replace(저장경로)

    정상특징[분할] = 결과
    print(f"{분할}: 저장 완료 {tuple(특징행렬.shape)}")
    print(저장경로)

print("\n정상 특징 저장 완료 — 모델 재학습·테스트 평가 없음")
```

```python
import numpy as np
import pandas as pd
import torch

from sklearn.neighbors import NearestNeighbors

학습특징 = 정상특징["train"]
검증특징 = 정상특징["validation"]

assert (
    학습특징["metadata"]["checkpoint_sha256"]
    == 검증특징["metadata"]["checkpoint_sha256"]
)

정상기준 = {
    "method": "same_chromosome_knn_mean_cosine_distance",
    "k": 5,
    "threshold_quantile": 0.95,
    "threshold_role": "정상 검증 자료에서 정한 임시 검토 기준",
    "feature_metadata": 학습특징["metadata"],
    "references": {},
}

검증기록 = []

for 번호 in ["chr9", "chr22"]:
    학습인덱스 = [
        i for i, 값 in enumerate(학습특징["chromosomes"])
        if 값 == 번호
    ]
    검증인덱스 = [
        i for i, 값 in enumerate(검증특징["chromosomes"])
        if 값 == 번호
    ]

    기준특징 = 학습특징["features"][학습인덱스].float()
    비교특징 = 검증특징["features"][검증인덱스].float()

    assert (기준특징.norm(dim=1) > 0).all().item()
    assert (비교특징.norm(dim=1) > 0).all().item()

    비교기 = NearestNeighbors(
        n_neighbors=5,
        metric="cosine",
        algorithm="brute",
    )
    비교기.fit(기준특징.numpy())

    거리, _ = 비교기.kneighbors(비교특징.numpy())
    점수들 = np.clip(거리, 0, 2).mean(axis=1)

    기준값 = float(np.quantile(
        점수들, 0.95, method="higher"
    ))

    정상기준["references"][번호] = {
        "features": 기준특징.clone(),
        "source_images": [
            학습특징["source_images"][i] for i in 학습인덱스
        ],
        "object_indices": [
            학습특징["object_indices"][i] for i in 학습인덱스
        ],
        "threshold": 기준값,
    }

    for 인덱스, 점수 in zip(검증인덱스, 점수들):
        검증기록.append({
            "source_image": 검증특징["source_images"][인덱스],
            "object_index": 검증특징["object_indices"][인덱스],
            "chromosome": 번호,
            "difference_score": float(점수),
            "threshold": 기준값,
            "review_candidate": bool(점수 > 기준값),
        })

    print(
        f"{번호} | 정상 학습 {len(학습인덱스)}개"
        f" | 검증 {len(검증인덱스)}개"
        f" | 임시 기준값 {기준값:.6f}"
    )

검증점수표 = pd.DataFrame(검증기록)

print("\n번호별 정상 검증 점수 분포:")
display(
    검증점수표.groupby("chromosome")["difference_score"]
    .describe()
)

print("\n정상 검증 자료 중 검토 후보 수:")
print(
    검증점수표.groupby("chromosome")["review_candidate"].sum()
)

print("\n기준 계산 완료 — 아직 저장·테스트 평가는 하지 않았습니다.")
```

```python
# 1. 현재 정상 기준 저장
기준경로 = 특징폴더 / "normal_knn5_cosine_v1.pt"

if 기준경로.exists():
    raise RuntimeError(
        "기준 파일이 이미 있습니다. 덮어쓰지 않았습니다. "
        "이 메시지를 알려주세요."
    )

임시경로 = 기준경로.with_suffix(".pt.part")
torch.save(정상기준, 임시경로)
임시경로.replace(기준경로)
print("정상 기준 저장 완료:", 기준경로)

# 저장된 기준을 다시 읽어 고정해서 사용
고정기준 = torch.load(
    기준경로, map_location="cpu", weights_only=True
)

# 2. 테스트 정상 조각 특징 추출
테스트목록경로 = 목록폴더 / "normal_chr9_chr22_test_v1.csv"
테스트표 = pd.read_csv(테스트목록경로).reset_index(drop=True)

assert set(테스트표["source_image"]).isdisjoint(
    학습특징["source_images"]
)
assert set(테스트표["source_image"]).isdisjoint(
    검증특징["source_images"]
)

전좌용_번호모델.eval()
테스트특징묶음 = []

with torch.inference_mode():
    for 시작 in tqdm(
        range(0, len(테스트표), 2),
        desc="테스트 정상 특징 추출",
    ):
        배치 = 테스트표.iloc[시작:시작 + 2]

        전체 = torch.stack([
            전좌용_이미지읽기(값)
            for 값 in 배치["global_crop_path"]
        ]).to(전좌용_장치)

        확대 = torch.stack([
            전좌용_이미지읽기(값)
            for 값 in 배치["local_crop_path"]
        ]).to(전좌용_장치)

        특징 = 전좌용_번호모델.extract_features(전체, 확대)
        assert torch.isfinite(특징).all().item()
        테스트특징묶음.append(특징.cpu())

테스트특징 = torch.cat(테스트특징묶음)
assert 테스트특징.shape == (len(테스트표), 2048)

# 3. 저장된 기준을 그대로 적용
테스트점수표 = 테스트표[
    ["source_image", "object_index", "chromosome"]
].copy()
테스트점수표["difference_score"] = np.nan
테스트점수표["threshold"] = np.nan

for 번호 in ["chr9", "chr22"]:
    선택 = 테스트표["chromosome"].eq(번호)
    인덱스 = np.flatnonzero(선택.to_numpy())
    기준 = 고정기준["references"][번호]

    비교기 = NearestNeighbors(
        n_neighbors=고정기준["k"],
        metric="cosine",
        algorithm="brute",
    )
    비교기.fit(기준["features"].numpy())

    거리, _ = 비교기.kneighbors(테스트특징[인덱스].numpy())
    점수 = np.clip(거리, 0, 2).mean(axis=1)

    테스트점수표.loc[선택, "difference_score"] = 점수
    테스트점수표.loc[선택, "threshold"] = 기준["threshold"]

assert 테스트점수표[
    ["difference_score", "threshold"]
].notna().all().all()

테스트점수표["review_candidate"] = (
    테스트점수표["difference_score"]
    > 테스트점수표["threshold"]
)

# 4. 결과와 특징 저장
결과폴더 = 프로젝트 / "philadelphia_data" / "reviews"
결과폴더.mkdir(parents=True, exist_ok=True)

테스트점수표.to_csv(
    결과폴더 / "normal_test_scores_v1.csv",
    index=False,
    encoding="utf-8-sig",
)

torch.save({
    "features": 테스트특징,
    "source_images": 테스트표["source_image"].tolist(),
    "object_indices": 테스트표["object_index"].tolist(),
    "chromosomes": 테스트표["chromosome"].tolist(),
    "checkpoint_sha256": 모델해시,
    "manifest_sha256": 파일해시(테스트목록경로),
    "preprocessing": str(전좌용_전처리),
}, 특징폴더 / "test_features.pt")

요약 = 테스트점수표.groupby("chromosome").agg(
    조각수=("review_candidate", "size"),
    검토후보수=("review_candidate", "sum"),
    기준초과율=("review_candidate", "mean"),
)
요약["기준초과율"] = 요약["기준초과율"].map(
    lambda 값: f"{값:.2%}"
)

display(요약)
print("테스트 확인 완료 — 기준값 변경·모델 재학습 없음")
```

```python
import numpy as np
from scipy import ndimage

# 기존 노트북과 같은 마스크 방식
def 참고용_마스크(rgb):
    회색 = np.asarray(
        Image.fromarray(rgb).convert("L"), dtype=np.float32
    )
    흐림 = ndimage.gaussian_filter(회색, sigma=0.6)
    값 = np.clip(흐림, 0, 255).astype(np.uint8)

    히스토그램 = np.bincount(값.ravel(), minlength=256).astype(float)
    확률 = 히스토그램 / 히스토그램.sum()
    누적비율 = np.cumsum(확률)
    누적평균 = np.cumsum(확률 * np.arange(256))

    분모 = 누적비율 * (1 - 누적비율)
    점수 = np.full(256, -1.0)
    유효 = 분모 > 0
    점수[유효] = (
        (누적평균[-1] * 누적비율[유효] - 누적평균[유효]) ** 2
        / 분모[유효]
    )
    전경 = 값 <= int(np.argmax(점수))

    덩어리, 개수 = ndimage.label(
        전경, structure=np.ones((3, 3), dtype=int)
    )
    if 개수 == 0:
        return np.zeros_like(전경)

    크기 = np.bincount(덩어리.ravel())
    크기[0] = 0
    return 덩어리 == 크기.argmax()


Wellcome_표 = 참고목록[
    참고목록["source_id"].eq("Wellcome_B0000302")
].reset_index(drop=True)

원본조각들 = []
for 경로 in Wellcome_표["crop_path"]:
    with Image.open(경로) as 사진:
        원본조각들.append(사진.convert("RGB"))

# 4개 조각 기준의 임시 바탕 — 전체 핵형 기준과 다름
임시바탕크기 = max(max(사진.size) for 사진 in 원본조각들) + 12
Wellcome_시험입력 = []

fig, axes = plt.subplots(4, 4, figsize=(12, 14))

for i, 원본조각 in enumerate(원본조각들):
    전체 = Image.new("RGB", (임시바탕크기, 임시바탕크기), "white")
    전체.paste(원본조각, (
        (임시바탕크기 - 원본조각.width) // 2,
        (임시바탕크기 - 원본조각.height) // 2,
    ))

    rgb = np.asarray(전체)
    마스크 = 참고용_마스크(rgb)
    정리배열 = rgb.copy()

    # 빈 마스크는 원본 유지
    if 마스크.any():
        정리배열[~마스크] = 255

    정리전체 = Image.fromarray(정리배열)
    ys, xs = np.where(마스크)

    if len(xs):
        x1, x2 = int(xs.min()), int(xs.max()) + 1
        y1, y2 = int(ys.min()), int(ys.max()) + 1
        여백 = max(2, int(np.ceil(max(x2-x1, y2-y1) * 0.08)))

        확대조각 = 정리전체.crop((
            max(0, x1 - 여백), max(0, y1 - 여백),
            min(임시바탕크기, x2 + 여백),
            min(임시바탕크기, y2 + 여백),
        ))
    else:
        확대조각 = 정리전체.copy()

    변길이 = max(확대조각.size)
    확대 = Image.new("RGB", (변길이, 변길이), "white")
    확대.paste(확대조각, (
        (변길이 - 확대조각.width) // 2,
        (변길이 - 확대조각.height) // 2,
    ))

    Wellcome_시험입력.append({
        "crop_path": Wellcome_표.iloc[i]["crop_path"],
        "global_image": 정리전체,
        "local_image": 확대,
        "scale_policy": "selected_four_crops_max_plus_12",
        "mask_empty": not bool(마스크.any()),
    })

    이름 = Path(Wellcome_표.iloc[i]["crop_path"]).stem
    for ax, 영상, 제목 in zip(axes[i], [
        원본조각, 마스크, 정리전체, 확대
    ], ["원본 조각", "선택된 마스크", "임시 전체 화면", "확대 화면"]):
        if 제목 == "선택된 마스크":
            ax.imshow(영상, cmap="gray", vmin=0, vmax=1)
        else:
            ax.imshow(영상)
        ax.set_title(제목)
        ax.axis("off")
    axes[i, 0].set_title(이름, fontsize=9)

plt.tight_layout()
plt.show()
print("임시 공통 바탕:", 임시바탕크기)
print("전처리 미리보기 완료 — 저장·점수 계산 없음")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import xml.etree.ElementTree as ET

from pathlib import Path
from PIL import Image

# 실험 v2: 정상·참고 사진에 공통으로 적용
def 공통v2_입력만들기(사진):
    사진 = 사진.convert("RGB")
    여백 = max(2, int(np.ceil(max(사진.size) * 0.08)))
    변길이 = max(사진.size) + 2 * 여백

    바탕 = Image.new("RGB", (변길이, 변길이), "white")
    바탕.paste(사진, (
        (변길이 - 사진.width) // 2,
        (변길이 - 사진.height) // 2,
    ))
    return 바탕


공통v2_사례캐시 = {}

def 공통v2_정상원본읽기(행, 분할):
    if 분할 not in 공통v2_사례캐시:
        사례표 = pd.read_csv(
            TVG루트 / "splits" / "tvg_number_652_v1"
            / f"{분할}_cases.csv"
        )
        assert 사례표["filename"].is_unique
        공통v2_사례캐시[분할] = 사례표.set_index("filename")

    사례 = 공통v2_사례캐시[분할].loc[행["source_image"]]
    assert 사례["source_category"] == "normal"

    XML = ET.parse(
        TVG루트 / 사례["annotation_path"]
    ).getroot()
    객체 = XML.findall("object")[int(행["object_index"])]

    # 객체 인덱스와 번호가 맞는지 확인
    예상라벨 = {"chr9": "C9", "chr22": "G22"}
    실제라벨 = (객체.findtext("name") or "").strip()
    assert 실제라벨 == 예상라벨[행["chromosome"]], (
        f"번호 불일치: {행['source_image']} / "
        f"{행['object_index']} / {실제라벨}"
    )

    x1, y1, x2, y2 = [
        int(객체.findtext(f"bndbox/{키}"))
        for 키 in ["xmin", "ymin", "xmax", "ymax"]
    ]

    with Image.open(TVG루트 / 사례["image_path"]) as 사진:
        영역 = (
            max(0, x1), max(0, y1),
            min(사진.width, x2), min(사진.height, y2),
        )
        assert 영역[2] > 영역[0] and 영역[3] > 영역[1]
        return 사진.convert("RGB").crop(영역)


# 정상 번호별 두 개씩 선택
학습표v2 = pd.read_csv(
    목록폴더 / "normal_chr9_chr22_train_v1.csv"
)
정상예시 = pd.concat([
    학습표v2[학습표v2["chromosome"].eq("chr9")].head(2),
    학습표v2[학습표v2["chromosome"].eq("chr22")].head(2),
], ignore_index=True)

미리보기v2 = []

for _, 행 in 정상예시.iterrows():
    조각 = 공통v2_정상원본읽기(행, "train")
    이름 = (
        f"TVG {행['chromosome']} | "
        f"{행['source_image']} / {행['object_index']}"
    )
    미리보기v2.append((이름, 조각))

for _, 행 in Wellcome_표.iterrows():
    with Image.open(행["crop_path"]) as 사진:
        조각 = 사진.convert("RGB")
    미리보기v2.append((Path(행["crop_path"]).stem, 조각))

fig, axes = plt.subplots(
    len(미리보기v2), 2,
    figsize=(8, 3 * len(미리보기v2)),
    squeeze=False,
)

for i, (이름, 조각) in enumerate(미리보기v2):
    共通入力 = 공통v2_입력만들기(조각)

    # 실제 모델에 들어가는 회색조 화면 확인
    입력텐서 = 전좌용_전처리(共通入力)
    평균 = 입력텐서.new_tensor(
        [0.485, 0.456, 0.406]
    )[:, None, None]
    표준편차 = 입력텐서.new_tensor(
        [0.229, 0.224, 0.225]
    )[:, None, None]
    표시영상 = (
        입력텐서 * 표준편차 + 평균
    ).clamp(0, 1).permute(1, 2, 0).numpy()

    axes[i, 0].imshow(조각)
    axes[i, 0].set_title(f"{이름}\n원본 조각", fontsize=9)

    axes[i, 1].imshow(표시영상)
    axes[i, 1].set_title("공통 v2 모델 입력 · 224×224")

    for ax in axes[i]:
        ax.axis("off")

plt.tight_layout()
plt.show()
print("공통 전처리 확인 — 학습·특징 추출·저장 없음")
```

```python
공통v2_특징폴더 = (
    프로젝트 / "philadelphia_data" / "features"
    / "finetuned652_single_preserved_v2"
)
공통v2_특징폴더.mkdir(parents=True, exist_ok=True)

공통v2_결과 = {}

작업들 = [
    (
        "train",
        목록폴더 / "normal_chr9_chr22_train_v1.csv",
        "normal",
    ),
    (
        "validation",
        목록폴더 / "normal_chr9_chr22_validation_v1.csv",
        "normal",
    ),
    (
        "wellcome",
        목록폴더 / "reference_crops_v1.csv",
        "reference",
    ),
]

전좌용_번호모델.eval()

for 이름, 목록경로, 종류 in 작업들:
    표 = pd.read_csv(목록경로)

    if 종류 == "reference":
        표 = 표[
            표["source_id"].eq("Wellcome_B0000302")
        ].copy()

    표 = 표.reset_index(drop=True)

    if 종류 == "normal":
        표본ID = [
            f"{행.source_image}#{int(행.object_index)}"
            for 행 in 표.itertuples()
        ]
        그룹ID = 표["source_image"].tolist()
    else:
        표본ID = [Path(값).name for 값 in 표["crop_path"]]
        그룹ID = 표["group_id"].tolist()

    assert len(표본ID) == len(set(표본ID))

    설정 = {
        "checkpoint_sha256": 모델해시,
        "manifest_sha256": 파일해시(목록경로),
        "input_method": "raw_bbox_no_mask_square_pad_8percent_min2",
        "preprocessing": str(전좌용_전처리),
        "feature_method": "finetuned_encoder_single_view",
        "feature_dimension": 1024,
        "augmentation": "none",
    }

    if 종류 == "normal":
        사례경로 = (
            TVG루트 / "splits" / "tvg_number_652_v1"
            / f"{이름}_cases.csv"
        )
        설정["case_manifest_sha256"] = 파일해시(사례경로)

    저장경로 = 공통v2_특징폴더 / f"{이름}_features.pt"

    if 저장경로.exists():
        결과 = torch.load(
            저장경로, map_location="cpu", weights_only=True
        )
        assert 결과["metadata"] == 설정, \
            f"{이름}: 기존 설정이 달라 덮어쓰지 않습니다."
        assert 결과["sample_ids"] == 표본ID
        assert 결과["group_ids"] == 그룹ID
        assert 결과["chromosomes"] == 표["chromosome"].tolist()
        assert 결과["features"].shape == (len(표), 1024)
        assert torch.isfinite(결과["features"]).all().item()

        공통v2_결과[이름] = 결과
        print(f"{이름}: 기존 특징 재사용")
        continue

    묶음 = []

    with torch.inference_mode():
        for 시작 in tqdm(
            range(0, len(표), 2),
            desc=f"v2 {이름} 특징 추출",
        ):
            입력들 = []

            for _, 행 in 표.iloc[시작:시작 + 2].iterrows():
                if 종류 == "normal":
                    조각 = 공통v2_정상원본읽기(행, 이름)
                else:
                    with Image.open(행["crop_path"]) as 사진:
                        조각 = 사진.convert("RGB")

                입력들.append(
                    전좌용_전처리(공통v2_입력만들기(조각))
                )

            배치 = torch.stack(입력들).to(전좌용_장치)

            # 두 화면 분류층 대신 공유 인코더만 사용
            특징 = 전좌용_번호모델.encoder(배치)
            assert torch.isfinite(특징).all().item()
            묶음.append(특징.cpu())

    행렬 = torch.cat(묶음)
    assert 행렬.shape == (len(표), 1024)

    결과 = {
        "features": 행렬,
        "sample_ids": 표본ID,
        "group_ids": 그룹ID,
        "chromosomes": 표["chromosome"].tolist(),
        "metadata": 설정,
    }

    임시경로 = 저장경로.with_suffix(".pt.part")
    torch.save(결과, 임시경로)
    임시경로.replace(저장경로)

    공통v2_결과[이름] = 결과
    print(f"{이름}: 저장 완료 {tuple(행렬.shape)}")

print("\nv2 특징 준비 완료 — 재학습·테스트 평가 없음")
```

```python
from sklearn.neighbors import NearestNeighbors

학습v2 = 공통v2_결과["train"]
검증v2 = 공통v2_결과["validation"]
참고v2 = 공통v2_결과["wellcome"]

assert set(학습v2["group_ids"]).isdisjoint(검증v2["group_ids"])

# 세 자료가 같은 모델과 전처리로 만들어졌는지 확인
for 결과 in [검증v2, 참고v2]:
    for 키 in [
        "checkpoint_sha256", "input_method", "preprocessing",
        "feature_method", "feature_dimension",
    ]:
        assert 결과["metadata"][키] == 학습v2["metadata"][키], 키

정상기준v2 = {
    "method": "same_chromosome_knn_mean_cosine_distance",
    "k": 5,
    "threshold_quantile": 0.95,
    "quantile_method": "higher",
    "feature_metadata": 학습v2["metadata"],
    "validation_manifest_sha256": 검증v2["metadata"]["manifest_sha256"],
    "references": {},
}

검증요약v2 = []
Wellcome기록v2 = []

for 번호 in ["chr9", "chr22"]:
    학습행 = np.flatnonzero(np.asarray(학습v2["chromosomes"]) == 번호)
    검증행 = np.flatnonzero(np.asarray(검증v2["chromosomes"]) == 번호)
    참고행 = np.flatnonzero(np.asarray(참고v2["chromosomes"]) == 번호)

    기준특징 = 학습v2["features"][학습행].float()
    비교기 = NearestNeighbors(
        n_neighbors=정상기준v2["k"],
        metric="cosine",
        algorithm="brute",
    ).fit(기준특징.numpy())

    검증거리, _ = 비교기.kneighbors(
        검증v2["features"][검증행].float().numpy()
    )
    검증점수 = np.clip(검증거리, 0, 2).mean(axis=1)
    기준값 = float(np.quantile(검증점수, 0.95, method="higher"))

    정상기준v2["references"][번호] = {
        "features": 기준특징,
        "sample_ids": [학습v2["sample_ids"][i] for i in 학습행],
        "threshold": 기준값,
    }
    검증요약v2.append({
        "번호": 번호,
        "학습 조각 수": len(학습행),
        "검증 조각 수": len(검증행),
        "검증 중앙값": float(np.median(검증점수)),
        "검증 최댓값": float(검증점수.max()),
        "임시 기준값": 기준값,
        "검증 기준 초과 수": int((검증점수 > 기준값).sum()),
    })

    참고거리, _ = 비교기.kneighbors(
        참고v2["features"][참고행].float().numpy()
    )
    참고점수 = np.clip(참고거리, 0, 2).mean(axis=1)

    for 행, 점수 in zip(참고행, 참고점수):
        Wellcome기록v2.append({
            "crop_file": 참고v2["sample_ids"][행],
            "chromosome": 번호,
            "reference_distance": float(점수),
            "tvg_validation_median": float(np.median(검증점수)),
            "tvg_validation_max": float(검증점수.max()),
            "usage": "exploratory_only",
        })

기준경로v2 = 공통v2_특징폴더 / "normal_knn5_cosine_v2.pt"
임시경로v2 = 기준경로v2.with_suffix(".pt.part")
torch.save(정상기준v2, 임시경로v2)
임시경로v2.replace(기준경로v2)

Wellcome점수표v2 = pd.DataFrame(Wellcome기록v2)
Wellcome점수표v2.to_csv(
    공통v2_특징폴더 / "wellcome_exploratory_scores_v2.csv",
    index=False,
    encoding="utf-8-sig",
)

display(pd.DataFrame(검증요약v2).round(6))
display(Wellcome점수표v2.round(6))
print("v2 정상 기준 저장:", 기준경로v2)
```

```python
고정기준v2 = torch.load(
    기준경로v2, map_location="cpu", weights_only=True
)
assert 고정기준v2["feature_metadata"]["checkpoint_sha256"] == 모델해시

테스트표v2 = pd.read_csv(
    목록폴더 / "normal_chr9_chr22_test_v1.csv"
).reset_index(drop=True)

테스트원본 = set(테스트표v2["source_image"])
for 분할 in ["train", "validation"]:
    assert 테스트원본.isdisjoint(공통v2_결과[분할]["group_ids"])

# 학습·검증 때와 동일한 v2 입력으로 특징 추출
테스트묶음v2 = []
전좌용_번호모델.eval()

with torch.inference_mode():
    for 시작 in tqdm(
        range(0, len(테스트표v2), 2),
        desc="v2 정상 테스트 특징 추출",
    ):
        입력들 = []
        for _, 행 in 테스트표v2.iloc[시작:시작 + 2].iterrows():
            조각 = 공통v2_정상원본읽기(행, "test")
            입력들.append(
                전좌용_전처리(공통v2_입력만들기(조각))
            )

        배치 = torch.stack(입력들).to(전좌용_장치)
        테스트묶음v2.append(
            전좌용_번호모델.encoder(배치).cpu()
        )

테스트특징v2 = torch.cat(테스트묶음v2)
assert 테스트특징v2.shape == (148, 1024)
assert torch.isfinite(테스트특징v2).all().item()

식별열 = ["source_image", "object_index", "chromosome"]
테스트점수표v2 = 테스트표v2[식별열].copy()
테스트점수표v2["difference_score"] = np.nan
테스트점수표v2["threshold"] = np.nan

for 번호 in ["chr9", "chr22"]:
    선택 = 테스트점수표v2["chromosome"].eq(번호)
    행번호 = np.flatnonzero(선택.to_numpy())
    기준 = 고정기준v2["references"][번호]

    비교기 = NearestNeighbors(
        n_neighbors=고정기준v2["k"],
        metric="cosine",
        algorithm="brute",
    ).fit(기준["features"].float().numpy())

    거리, _ = 비교기.kneighbors(
        테스트특징v2[행번호].float().numpy()
    )
    테스트점수표v2.loc[선택, "difference_score"] = (
        np.clip(거리, 0, 2).mean(axis=1)
    )
    테스트점수표v2.loc[선택, "threshold"] = 기준["threshold"]

assert 테스트점수표v2[
    ["difference_score", "threshold"]
].notna().all().all()

테스트점수표v2["review_candidate"] = (
    테스트점수표v2["difference_score"]
    > 테스트점수표v2["threshold"]
)

# 원본 파일명·객체 번호·염색체 번호로 동일한 조각끼리 연결
검토폴더 = 프로젝트 / "philadelphia_data" / "reviews"
v1점수표 = pd.read_csv(검토폴더 / "normal_test_scores_v1.csv")

비교표 = v1점수표[식별열 + ["review_candidate"]].merge(
    테스트점수표v2[식별열 + ["review_candidate"]],
    on=식별열,
    how="inner",
    suffixes=("_v1", "_v2"),
    validate="one_to_one",
)
assert len(비교표) == len(v1점수표) == len(테스트점수표v2) == 148

요약기록 = []
for 번호, 표 in list(비교표.groupby("chromosome")) + [("전체", 비교표)]:
    요약기록.append({
        "번호": 번호,
        "조각 수": len(표),
        "v1 후보 수": int(표["review_candidate_v1"].sum()),
        "v1 기준 초과율": f"{표['review_candidate_v1'].mean():.2%}",
        "v2 후보 수": int(표["review_candidate_v2"].sum()),
        "v2 기준 초과율": f"{표['review_candidate_v2'].mean():.2%}",
    })

저장경로v2 = 검토폴더 / "normal_test_scores_v2.csv"
테스트점수표v2.to_csv(저장경로v2, index=False, encoding="utf-8-sig")

display(pd.DataFrame(요약기록))
print("v2 정상 테스트 결과 저장:", 저장경로v2)
```

```python
T922_표v2 = 참고목록.loc[
    참고목록["source_id"].eq("T922_CML")
].reset_index(drop=True)

assert len(T922_표v2) == 4

입력들 = []
for 경로 in T922_표v2["crop_path"]:
    with Image.open(경로) as 사진:
        입력들.append(
            전좌용_전처리(공통v2_입력만들기(사진))
        )

전좌용_번호모델.eval()
with torch.inference_mode():
    배치 = torch.stack(입력들).to(전좌용_장치)
    T922_특징v2 = 전좌용_번호모델.encoder(배치).cpu().float()

assert T922_특징v2.shape == (4, 1024)
assert torch.isfinite(T922_특징v2).all().item()

T922_기록v2 = []

for 번호 in ["chr9", "chr22"]:
    선택행 = np.flatnonzero(
        T922_표v2["chromosome"].eq(번호).to_numpy()
    )
    조각표 = T922_표v2.iloc[선택행]

    assert len(조각표) == 2
    assert set(조각표["status"]) == {"정상 상동", "파생"}

    기준 = 고정기준v2["references"][번호]
    비교기 = NearestNeighbors(
        n_neighbors=고정기준v2["k"],
        metric="cosine",
        algorithm="brute",
    ).fit(기준["features"].float().numpy())

    거리, _ = 비교기.kneighbors(
        T922_특징v2[선택행].numpy()
    )

    for (_, 조각), 점수 in zip(
        조각표.iterrows(), np.clip(거리, 0, 2).mean(axis=1)
    ):
        T922_기록v2.append({
            "crop_file": Path(조각["crop_path"]).name,
            "chromosome": 번호,
            "source_label": 조각["status"],
            "reference_distance": float(점수),
        })

T922_결과v2 = pd.DataFrame(T922_기록v2)
T922_비교v2 = T922_결과v2.pivot(
    index="chromosome",
    columns="source_label",
    values="reference_distance",
)
T922_비교v2["파생 − 정상 상동"] = (
    T922_비교v2["파생"] - T922_비교v2["정상 상동"]
)

실험폴더 = 프로젝트 / "philadelphia_data" / "experiments"
실험폴더.mkdir(parents=True, exist_ok=True)
T922_결과v2.to_csv(
    실험폴더 / "t922_paired_exploratory_scores_v2.csv",
    index=False,
    encoding="utf-8-sig",
)

display(T922_비교v2[["정상 상동", "파생", "파생 − 정상 상동"]].round(6))
```
